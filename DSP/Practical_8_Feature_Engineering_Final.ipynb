## Step 1: Import Required Libraries

```python
print("[REDACTED] [REDACTED]")
# NEW: Required visualization libraries are imported here.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
```

## Step 2: Load the Iris Dataset

```python
print("[REDACTED] [REDACTED]")
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = iris.target
# NEW: Add readable species labels for visualization.
df["species_name"] = df["species"].map(dict(enumerate(iris.target_names)))

print("Iris dataset loaded successfully.")
display(df.head())
```

## Step 3: Display Dataset Size, Shape and Information

```python
print("[REDACTED] [REDACTED]")
# NEW SECTION: Display dataset size, shape, information, and column names.
print("Dataset size (number of cells):", df.size)
print("Dataset shape (rows, columns):", df.shape)
print("\nDataset information:")
df.info()
print("\nColumn names:")
print(df.columns)
```

## Step 4: Check Missing Values

```python
print("[REDACTED] [REDACTED]")
# NEW SECTION: Check for missing values.
print("Missing values in each column:")
print(df.isnull().sum())
```

## Step 5: Visualize the Dataset Using Seaborn

```python
print("[REDACTED] [REDACTED]")
# NEW SECTION: Visualize the original Iris measurements before feature engineering.
sns.pairplot(df, vars=iris.feature_names, hue="species_name", diag_kind="hist")
plt.show()
```

The pair plot compares the four original Iris measurements. `hue="species_name"` colours points by species.

## Step 6: Create a New Feature

```python
print("[REDACTED] [REDACTED]")
# Existing required feature; retained from the original practical.
df["petal_area"] = df["petal length (cm)"] * df["petal width (cm)"]
print("Dataset with the new petal_area feature:")
display(df.head())
```

`petal_area` is calculated by multiplying petal length by petal width. The earlier `sepal_ratio` feature has been removed to keep the code aligned with the specified aim.

## Step 7: Transform Features Using StandardScaler

```python
print("[REDACTED] [REDACTED]")
# CHANGED: Scale only numerical features, including the new petal_area feature.
feature_columns = df[iris.feature_names + ["petal_area"]]
scaler = StandardScaler()
scaled_values = scaler.fit_transform(feature_columns)
scaled_df = pd.DataFrame(scaled_values, columns=feature_columns.columns)

print("Standardized features (first five rows):")
display(scaled_df.head())
```

`StandardScaler()` standardizes each feature to approximately zero mean and unit variance. `fit_transform()` learns the scaling parameters and transforms the features.

## Step 8: Select Relevant Features

```python
print("[REDACTED] [REDACTED]")
# CHANGED: Select features from the original measurements plus petal_area.
X = df[iris.feature_names + ["petal_area"]]
y = df["species"]

selector = SelectKBest(score_func=f_classif, k=3)
selected_values = selector.fit_transform(X, y)
selected_features = X.columns[selector.get_support()]

print("Selected features:")
print(list(selected_features))
print("\nSelected feature data (first five rows):")
display(pd.DataFrame(selected_values, columns=selected_features).head())
```

`SelectKBest` uses the ANOVA F-test (`f_classif`) to select the three features with the strongest statistical relationship to the species labels.

## Step 9: Visualize Selected Features

```python
print("[REDACTED] [REDACTED]")
sns.scatterplot(data=df, x=selected_features[0], y=selected_features[1], hue="species_name")
plt.title("Selected Features of the Iris Dataset")
plt.tight_layout()
plt.show()
```

## Step 10: Display Correlation Heatmap

```python
print("[REDACTED] [REDACTED]")
# NEW SECTION: Calculate and display the feature correlation heatmap.
correlation = df[iris.feature_names + ["petal_area"]].corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Iris Features")
plt.tight_layout()
plt.show()
```

The heatmap shows pairwise correlation coefficients between the original numerical measurements and the engineered `petal_area` feature.

The Iris dataset was loaded and inspected, missing values were checked, and the original data was visualized. A new `petal_area` feature was created, features were standardized and selected, and the selected features and correlation heatmap were visualized.
