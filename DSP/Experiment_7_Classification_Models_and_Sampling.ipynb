# Experiment 7 — Evaluating Machine Learning Classification Models and Sampling

**Objective:**  
To perform sampling for classification and evaluate different machine learning classification models using the Iris dataset.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedShuffleSplit
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# Print student details as the first output line
print("[REDACTED] [REDACTED]")

# Load the Iris dataset
df = sns.load_dataset("iris")

print("\nFirst five records:")
display(df.head())
```

## 1. Prepare Data for Classification

```python
# Separate input features and target variable
X = df.drop("species", axis=1)
y = df["species"]

print("Features:")
display(X.head())

print("\nTarget:")
display(y.head())

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)

# Display class distribution
print("\nClass distribution:")
print(y.value_counts())
```

## 2. Sampling for Classification

```python
# Perform a stratified train-test split
# Stratification preserves the proportion of each class in both sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
# Verify the proportion of each class before and after sampling
print("Original class proportions:")
print(y.value_counts(normalize=True))

print("\nTraining class proportions:")
print(y_train.value_counts(normalize=True))

print("\nTesting class proportions:")
print(y_test.value_counts(normalize=True))
```

```python
# Demonstrate stratified sampling using StratifiedShuffleSplit
splitter = StratifiedShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

for train_index, test_index in splitter.split(X, y):
    X_train_sample = X.iloc[train_index]
    X_test_sample = X.iloc[test_index]
    y_train_sample = y.iloc[train_index]
    y_test_sample = y.iloc[test_index]

print("Stratified sample training size:", len(X_train_sample))
print("Stratified sample testing size:", len(X_test_sample))
```

## 3. Train Classification Models

```python
# Create the classification models
models = {
    "Logistic Regression": LogisticRegression(max_iter=200),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    )
}

# Train each model
for name, model in models.items():
    model.fit(X_train, y_train)
    print(name, "trained successfully.")
```

## 4. Evaluate the Classification Models

```python
# Generate predictions and calculate accuracy
results = []

for name, model in models.items():
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)

    results.append({
        "Model": name,
        "Accuracy": accuracy
    })

results_df = pd.DataFrame(results)

print("Model evaluation results:")
display(results_df)
```

```python
# Display confusion matrices for the classification models
for name, model in models.items():
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=model.classes_,
        yticklabels=model.classes_
    )
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.show()
```

## 5. Compare the Classification Models

```python
# Sort models according to their accuracy
results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

print("Classification model comparison:")
display(results_df)

# Visualize model accuracy
plt.figure(figsize=(8, 5))

plt.bar(
    results_df["Model"],
    results_df["Accuracy"]
)

plt.title("Classification Model Accuracy")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1.05)
plt.xticks(rotation=15)

plt.show()
```

## Conclusion

The Iris dataset was prepared for classification and stratified sampling was performed to preserve the class distribution in the training and testing sets. Logistic Regression, Decision Tree, and Random Forest classification models were trained and evaluated using the test data. Their accuracies were compared and confusion matrices were generated to examine the classification results.
